```python
import pandas as pd
from statsmodels.graphics.factorplots import interaction_plot
```

**interaction** : It means whether the levels of one factor affects the level of another factor.

```python
insurance_df = pd.read_csv('insurance.csv', index_col='Unnamed: 0') 
insurance_df.head()
```

ValueError: '.' is not a valid value for ls; supported values are '-', '--', '-.', ':', 'None', ' ', '', 'solid', 'dashed', 'dashdot', 'dotted'

```python
insurance_df.groupby('smoker')['charges'].mean()
```

```python
fig = interaction_plot(x=insurance_df.smoker,
                       trace=insurance_df.sex,
                       response=insurance_df.charges,
                       plottype='b',
                       colors=['pink','black'],
                       markers=['^','^'],
                       func='mean',
                       ylabel='medical charges')
```

Q: Is there a main effect between sex?

A: The difference b/w the charges of male and female is almost 0. Hence, we can say that there is not a major effect b/w these two.

Q: Is there a main effect between smoker(y/n)?

A: The patients who are smoker tends to have very high surge in medical charges as compared to non-smoker patients, which seem obvious since smoker patients are harder to treat.
